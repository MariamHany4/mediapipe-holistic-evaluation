```python
import os
import cv2
import re
from collections import defaultdict

BASE_DIR = "."
VIDEOS_ROOT = os.path.join(BASE_DIR, "videos")
OUTPUT_DIR = os.path.join(BASE_DIR, "extracted-frames")

BATCHES = [
    "frames-batch-1",
    "frames-batch-2",
    "frames-batch-3",
    "frames-batch-4"
]

pattern = re.compile(r"(.*)_frame(\d+)\.")

def process_batch(batch_name):
    print(f"\nProcessing {batch_name}...")

    # frames path
    frames_dir = os.path.join(BASE_DIR, batch_name, "frames")

    # map batch name → videos subfolder
    batch_number = batch_name.split("-")[-1]   # "1"
    videos_dir = os.path.join(VIDEOS_ROOT, f"batch {batch_number}")

    # output path
    output_batch_dir = os.path.join(OUTPUT_DIR, f"batch-{batch_number}")
    os.makedirs(output_batch_dir, exist_ok=True)

    # sanity checks
    if not os.path.exists(frames_dir):
        print(f"Frames folder not found: {frames_dir}")
        return

    if not os.path.exists(videos_dir):
        print(f"Videos folder not found: {videos_dir}")
        return

    # Step 1: Collect required frames
    video_frames = defaultdict(list)

    for file in os.listdir(frames_dir):
        match = pattern.match(file)
        if match:
            video_name = match.group(1)
            frame_num = int(match.group(2))
            video_frames[video_name].append(frame_num)

    # Step 2: Extract frames
    for video_name, frame_list in video_frames.items():
        video_path = os.path.join(videos_dir, f"{video_name}.mp4")

        if not os.path.exists(video_path):
            print(f"Video not found: {video_path}")
            continue

        cap = cv2.VideoCapture(video_path)

        if not cap.isOpened():
            print(f"Failed to open: {video_name}")
            continue

        frame_set = set(frame_list)
        max_frame = max(frame_set)

        current_frame = 0
        saved = 0

        while current_frame <= max_frame:
            ret, frame = cap.read()
            if not ret:
                break

            if current_frame in frame_set:
                out_name = f"{video_name}_frame{current_frame:04d}.jpg"
                out_path = os.path.join(output_batch_dir, out_name)
                cv2.imwrite(out_path, frame)
                saved += 1

            current_frame += 1

        cap.release()
        print(f"{video_name}: saved {saved} frames")


# Run all batches
for batch in BATCHES:
    process_batch(batch)

print("\nDone!")
```
