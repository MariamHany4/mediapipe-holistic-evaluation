```python
import os
import cv2
```

```python
VIDEO_FILE = "25787566993349564-CROSS 2.mp4"    
FRAMES_DIR = "vid frames"     
os.makedirs(FRAMES_DIR, exist_ok=True)
```

```python
# Open the video
cap = cv2.VideoCapture(VIDEO_FILE)
if not cap.isOpened():
    print(f"Failed to open {VIDEO_FILE}")
    exit()

frame_count = 0
video_name = os.path.splitext(os.path.basename(VIDEO_FILE))[0]

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Save frame with unique name
    frame_filename = f"{video_name}_frame{frame_count:04d}.jpg"
    frame_path = os.path.join(FRAMES_DIR, frame_filename)
    cv2.imwrite(frame_path, frame)

    frame_count += 1

cap.release()
print(f"Extracted {frame_count} frames from {VIDEO_FILE}")
```
